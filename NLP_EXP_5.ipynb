```python
import nltk
from nltk import word_tokenize, bigrams, FreqDist, ConditionalFreqDist
from nltk.corpus import stopwords
from collections import defaultdict

nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
def preprocess(text):
  text = text.lower()
  tokens = word_tokenize(text)
  stop_words = set(stopwords.words('english'))
  filtered_tokens = [token for token in tokens if token.isalnum() and token not in stop_words]
  return filtered_tokens
```

```python
corpus = "Today we are learning N-gram model implementation. It is a very intresting topic."
```

```python
filtered_tokens = preprocess(corpus)
filtered_tokens
```

```python
bigrams_list = list(bigrams(filtered_tokens))
bigrams_list
```

```python
bigram_freq = FreqDist(bigrams_list)
bigram_freq
```

```python
cfd = ConditionalFreqDist(bigrams_list)

cfd.items()
```

```python
from nltk import trigrams
trigrams_list = list(trigrams(filtered_tokens))
trigrams_list
```

```python
def predict_next_word(word):
  word = word.lower()
  if word in cfd:
    next_word = cfd[word].max()
    return next_word
  else:
    return None
```

```python
start_word = "today"
predicted_word = predict_next_word(start_word)
if predicted_word:
  print(f"Next word after '{start_word}' is '{predicted_word}'.")
else:
  print(f"No bigram found for '{start_word}'.")
```

```python
def generate_sequence(start_word, num_words=8):
  sequence = [start_word]
  current_word = start_word
  for _ in range(num_words):
    next_word = predict_next_word(current_word)
    if next_word:
      sequence.append(next_word)
      current_word = next_word
    else:
      break
  return " ".join(sequence)

generated_text = generate_sequence("today", num_words=8)
print(f"Generated sequence: {generated_text}")
```
