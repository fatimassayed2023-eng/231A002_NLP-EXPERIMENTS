```python
import nltk

nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("maxent_ne_chunker_tab")
nltk.download("words")
```

```python
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk
```

```python
text = "I hope this message finds you well, Monica. As discussed in our last meeting, Netflix is ready to move forward with the software integration project in collaboration with Harper & Cole Consulting."

tokens = word_tokenize(text)
print(tokens)
```

```python
tagged_tokens = pos_tag(tokens)
print(tagged_tokens)
```

```python
nltk.download('tagsets_json')
nltk.help.upenn_tagset("VBG")
```

```python
pos_tag(["are"])
```

```python
named_entities = ne_chunk(tagged_tokens)
print(named_entities)
```

```python
for subtree in named_entities:
    if isinstance(subtree, nltk.Tree) :
        entity = " ".join([word for word, tag in subtree.leaves()])
        label = subtree.label()
        print(f"Entity:",entity)
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text="I hope this message finds you well, Monica. As discussed in our last meeting, NETFLIX is ready to move forward with the software integration project in collaboration with Harper & Cole Consulting."
doc = nlp(text)
print(doc)


for ent in doc.ents:
    print(f"Entity: {ent.text}, Label: {ent.label_}")
```

PART 1 - CHUNKING

```python
import nltk
from nltk import word_tokenize, pos_tag
from nltk.chunk import RegexpParser
```

```python
sent = "The smart girl is also very good at Badminton"
```

```python
tokens = word_tokenize(sent)
tagged = pos_tag(tokens)
print(tagged)
```

```python
grammar = "NP: {<DT>?<JJ>*<NN>}"
cp = RegexpParser(grammar)
result = cp.parse(tagged)
print(result)
```
