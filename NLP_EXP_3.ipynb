```python
print("Fatima Sayed - 231A002\n")
text_lower = (text.lower())
print(text_lower)
```

```python
print("Fatima Sayed - 231A002\n")
import re
pattern = r'https?://www\.\S+|www\.\S+|\S\@\S+|\@\S+|\#\S+|\d\%|\d\S+'
text_url=re.sub(pattern," ",text_lower)
print(text_url)
```

```python
print("Fatima Sayed - 231A002\n")
!pip install emoji

import emoji
```

```python
print("Fatima Sayed - 231A002\n")
text_emoji = emoji.replace_emoji(text_url,replace="")
print(text_emoji)
```

```python
print("Fatima Sayed - 231A002\n")
import nltk
nltk.download('punkt_tab')
```

```python
print("Fatima Sayed - 231A002\n")
from nltk import word_tokenize
tokenized = word_tokenize(text_emoji)
print("Tokenized words:" , tokenized)
```

```python
print("Fatima Sayed - 231A002\n")
import string
punctuation = list(string.punctuation)
punctuation.extend(["``", "''"])
print(punctuation)
```

```python
print("Fatima Sayed - 231A002\n")
text_clean=[]
for word in tokenized:
  if word not in punctuation: # Use the corrected 'punctuation' list
    text_clean.append(word)
print(text_clean)
```

```python
print("Fatima Sayed - 231A002\n")
print("With punctuation:",tokenized)
print(len(tokenized))

print("Without punctuation:",text_clean)
print(len(text_clean))
```

```python
print("Fatima Sayed - 231A002\n")
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
stop_words = stopwords.words('english')
```

```python
print(stop_words)
```

```python
print("Fatima Sayed - 231A002\n")

from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
text_stemmed = [stemmer.stem(word) for word in text_clean]
print(text_stemmed)
```

```python
print("Fatima Sayed - 231A002\n")

import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
text_clean4=[]
lemma = WordNetLemmatizer()
text_lemmatized=[lemma.lemmatize(word) for word in text_clean]

print("Lemmatized text:" ,text_lemmatized)
print()
print("Stemmed text:", text_stemmed)

```
